```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("Games_sales.csv")
data
```

```python
data.head()
```

```python
data.info()
```

```python
data=data.dropna()
```

```python
data['Year']=data['Year'].astype(int)
```

```python
data.info()
```

```python
data.head()
```

```python
data.describe()
```

```python
Top_Games=data.groupby('Name')['Global_Sales'].sum().sort_values(ascending=False).head(10)
Top_Games
```

```python
Top_Games.plot(kind='bar')
plt.title("Top 10 Best Selling Games")
plt.xlabel("Game")
plt.ylabel("Global Sales")
plt.show()
```

```python
Top_Platform=data.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False).head(10)
Top_Platform
```

```python
Top_Platform.plot(kind='line')
plt.title("Top Platforms")
plt.ylabel("Global Sales")
plt.show()
```

```python
Genre_Sales=data.groupby('Genre')['Global_Sales'].sum().sort_values(ascending=False).head(10)
Genre_Sales
```

```python
Genre_Sales.plot(kind='pie')
plt.title("Genre Share")
plt.ylabel("")
plt.show()
```

```python
Year_Sales=data.groupby('Year')['Global_Sales'].sum().sort_values(ascending=False).head(10)
Year_Sales
```

```python
Year_Sales.plot(kind='bar')
plt.show()
```

```python
regional = data[['NA_Sales','EU_Sales','JP_Sales','Other_Sales']].sum()
regional.plot(kind='bar')
plt.title("Regional Sales Comparison")
plt.show()
```

```python
jp_genre = data.groupby('Genre')['JP_Sales'].sum().sort_values(ascending=False)
jp_genre.plot(kind='bar')
plt.title("Japan Genre Preference")
plt.show()
```
